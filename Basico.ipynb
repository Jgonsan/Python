```python
#Operações matématica de união, interseção e diferença

meu_set = {1,2,3,4,5,8,9}
meu_set2 = {1,4,5,6,8,11,12}

#união

uniao1 = meu_set | meu_set2
uniao2 = meu_set.union(meu_set2)

print(uniao1)
print(uniao2)

#Interseção

intersecao1 = meu_set & meu_set2
intersecao2 = meu_set.intersection(meu_set2)

print(intersecao1)
print(intersecao2)

#Diferença

diferenca1 = meu_set - meu_set2
diferenca2 = meu_set.difference(meu_set2)

print(diferenca1)
print(diferenca2)
```

```python
#Função que calcula a rentabilidade

def calcula_rentabilidade(valor_inicial, valor_final):
    rentabilidade = valor_final/valor_inicial - 1
    rentabilidade_em_porcentagem = '{:.2%}'.format(rentabilidade)
    return rentabilidade_em_porcentagem

valor_inicial = float(input('Digite a cotação inicial: '))
valor_final = float(input('Digite a cotação final: '))
rentabilidade = calcula_rentabilidade(valor_inicial=valor_inicial, valor_final=valor_final)

print('A rentabilidade  foi de =>', rentabilidade)

```

```python
#Desafio

continuar = True
lista_cotacao = []

while continuar:
    
    cotacao = float(input('Digite a cotação:'))
    lista_cotacao.append(cotacao)
    
    deseja_continuar = str(input('Deseja continuar! [S/N]')).lower()
    
    if deseja_continuar != 'n' and deseja_continuar != 's':
        invalido = True

        while invalido:
            deseja_continuar = str(input('Por favor, digite um comando válido: [S/N]')).lower()

            if deseja_continuar == 'n' or deseja_continuar == 's':
                break

    if deseja_continuar == 'n' and len(lista_cotacao) > 1:
        break

    elif deseja_continuar == 'n' and len(lista_cotacao) == 1:

        print('Não é possivel calcular retornos com apenas uma cotação. Por favor, digite mais uma.')

dicionario_retornos = {}

def calcula_retornos(lista_cotacao):
    for i in range(1, len(lista_cotacao)):
        retorno = lista_cotacao[i] / lista_cotacao[i-1] - 1
        retorno_percentual = '{:.2%}'.format(retorno)
        dicionario_retornos[f'Retorno dia {i}'] = retorno_percentual 

calcula_retornos(lista_cotacao=lista_cotacao)
        
print(dicionario_retornos)

```

### Tratamento de erros

```python
#Calculadora

numero1 = float(input('Digite um número: '))
numero2 = float(input('Digite outro número: '))

divisao = numero1 / numero2

print(f'A divisão entre o primeiro e o segundo número é: {divisao}')
```

```python
#Calculadora 2

try:

    numero1 = float(input('Digite um número: '))
    numero2 = float(input('Digite outro número: '))

    divisao = numero1 / numero2

except ZeroDivisionError:
    print('Não é possível dividir um número por 0!')

except ValueError:
    print('Ocorreu um erro com o tipo de dados digitado.')

else:

    print(f'A divisão entre o primeiro e o segundo número é: {divisao}')

finally: 

    print('Obrigado por participar')
```

```python
#Calculadora 3

try:

    numero1 = float(input('Digite um número: '))
    numero2 = float(input('Digite outro número: '))

    divisao = numero1 / numero2

except Exception as erro:
    print(f'Ocorreu um erro do tipo {erro.__class__}')

else:

    print(f'A divisão entre o primeiro e o segundo número é: {divisao}')

finally: 

    print('Obrigado por participar')
```

```python
#Extraindo informações de datas

lista_datas = [date(2022, 8, 31), date(2021, 5, 14), date(2020, 9, 21)]

for data in lista_datas:
    print(data.year)
    print(data.month)
    print(data.day)
```
