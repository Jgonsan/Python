### Propriedades de operações geométricas

```python
import pandas as pd
import numpy as np
```

```python
#Definição de uma função para gerar numeros aleatorios

def gerador_serie_historica(valor_inicial, volatilidade, periodos, dia_inicial, frequencia='M'):
    vetor = [valor_inicial]
    for i in range(periodos -1):
            preco = vetor[i] * (1 + np.random.normal(0, volatilidade))
            vetor.append(preco)
    serie = pd.Series(vetor, index=pd.date_range(start=dia_inicial, periods=periodos, freq=frequencia))
    return serie


```

```python
# Operação de multiplicação entre duas séries com índices diferentes
serie_euro = gerador_serie_historica(5, 0.01, 5, '2022-01-01')
serie_cotacoes = gerador_serie_historica(20, 0.05, 5, '2022-01-01')

print(serie_euro)
print(serie_cotacoes)
```

```python
# Operação de multiplicação entre duas séries com índices diferentes

final = serie_euro * serie_cotacoes
print(final)
```

```python
# Operação de divisão entre duas séries com índices diferentes

final = serie_euro / serie_cotacoes

print(final)
```

```python
# Operação de soma entre duas séries com índices diferentes

final = serie_euro + serie_cotacoes

print(final)
```

```python
# Operação de subtração entre duas séries com índices diferentes
final = serie_euro - serie_cotacoes

print(final)
```

```python
# Operação exponenciação entre duas séries com índices diferentes

final = serie_euro ** serie_cotacoes

print(final)    
```

```python
#Criando um DataFrame com duas séries
serie_lucro = gerador_serie_historica(2, 0.5, 5, '2022-01-01')

df = pd.DataFrame({'cotacao': serie_cotacoes, 'lucro_por_acao': serie_lucro})

df
```

```python
# Criando uma coluna com o Preço sobre lucro (P/L)
df['P/L'] = df['cotacao'] / df['lucro_por_acao']
df
```

```python
# Operação aritmeticas em colunas de DataFrame
#.sum(), .mean(), .max(), .min(), .std(), .var()

serie_lucro_trimestral = gerador_serie_historica(20000, 2, 4, '2022-01-01', frequencia='QE')
serie_lucro_trimestral

```

```python
#sum() - soma dos valores
Lucro_anual = serie_lucro_trimestral.sum()
print(Lucro_anual)
```

```python
#cumsum() - soma cumulativa
crescimento_lucro = serie_lucro_trimestral.cumsum()
print(crescimento_lucro)
```

```python
#max() - valor máximo
lucro_maximo = serie_lucro_trimestral.max()
print(lucro_maximo)
```

```python
#min() - valor mínimo
lucro_minimo = serie_lucro_trimestral.min()
print(lucro_minimo)
```

### Formas de ordenação de dados

```python
# sort_index() - ordena o DataFrame pelo índice

dicionario = {
    'nomes': ['Quero Quero', 'Alpagartas', 'Alpagartas', 'Magazine Luiza'],
    'precos_sobre_lucro': [12, 6, 12, 100],
    'volumes': [5000, 1000, 4000, 7000]}

empresas = pd.DataFrame(dicionario, index=['LJQQ3', 'ALPA4', 'ALPA3', 'MGLU3'])
empresas =  empresas.sort_index()
print(empresas)
empresas = empresas.sort_index(ascending=False)
print(empresas)
empresas = empresas.sort_index(axis=1, ascending=False)
print(empresas)
```

```python
#sort_values() - ordena o DataFrame pelos valores de uma coluna específica

acoes = {'Acoes':['Wege', 'Petrobras', 'Vale', 'PetroRio'],
         'Valor': [35.42, 33.67, 72.22, 33.96]}
df_acoes = pd.DataFrame(acoes, index=[1,2,3,4])
print(df_acoes)
         
```

```python
df_acoes = df_acoes.sort_values(by='Valor', ascending=False)
print(df_acoes)
```

```python
#rank() - atribui uma classificação aos valores de uma coluna específica
df_acoes['Rank'] = df_acoes['Valor'].rank(ascending=False)
print(df_acoes)
```

```python
empresas = empresas.rank(method='dense', numeric_only=True, na_option='keep', ascending=False)
print(empresas)
```

```python
#Tipos de rank

empresas['rank_min'] = empresas['precos_sobre_lucro'].rank(method='min')
empresas['rank_max'] = empresas['precos_sobre_lucro'].rank(method='max')
empresas['rank_first'] = empresas['precos_sobre_lucro'].rank(method='first')
empresas['rank_dense'] = empresas['precos_sobre_lucro'].rank(method='dense')
empresas
```
