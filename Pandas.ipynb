```python
#Desafio 2
import random
import matplotlib.pyplot as plt

cotacao_inicial = float(input('Informe a cotação inicial da ação: '))
numero_dias = int(input('Informe o número de dias para simulação: '))
lista_cotacoes = [cotacao_inicial]
lista_dias = [0]
volatilidade = 0.01  # 1% de volatilidade diária

for i in range(1, numero_dias + 1):
    # Gera um retorno diário aleatório baseado na volatilidade
    retorno_diario = np.random.normal(0, volatilidade)
    nova_cotacao = lista_cotacoes[-1] * (1 + retorno_diario)
    lista_cotacoes.append(nova_cotacao)
    lista_dias.append(i)
    print(f'Dia {i}: Cotação = {nova_cotacao:.2f}')


print(lista_cotacoes)

plt.plot(lista_cotacoes)

```

### Propriedades de seleção de dados dentro dos DataFrames

```python
#loc - Seleciona linhas e colunas por rótulo
df_cambio = pd.DataFrame({
    'moeda': ['Dolar/real', 'Libra/real']*4,
    'price': [5.2, 6.1, 7.0, 5.3, 6.2, 7.1, 5.4, 6.3],
    'derivativo': list(['DOLFT3', 'LIBFT3'])*4},
    index = sorted(list(pd.date_range('1/5/2014',periods=4))*2))

df_cambio
```

```python
#um unico dia por chave

print(df_cambio.loc['2014-01-05'])
```

```python
#selecionar um intervalo

df_cambio.loc['2014-01-05':'2014-01-07']
```

```python
#Selecionando uma chave e limitando o intervalo da coluna

df_cambio.loc['2014-01-05','moeda':'price']
```

```python
#Substituir dados com loc:

df_cambio.loc[df_cambio['derivativo'] == 'DOLFT3', 'derivativo'] = 'DOLFT4'
print(df_cambio)
```

```python
#at - Seleciona linhas e colunas por posição

print(df_cambio.at['2014-01-05', 'price'])
```

```python
#iloc - Seleciona linhas e colunas por posição
print(df_cambio.iloc[0])
```

```python
#intervalo de linhas
print(df_cambio.iloc[0:3])

#intervalo de colunas
print(df_cambio.iloc[:, 0])

#selecionando apenas um item
print(df_cambio.iloc[0, 1])
```

```python
#mudando o valor do item

df_cambio.iloc[0,0] = 99 
print(df_cambio.iloc[0])
```
