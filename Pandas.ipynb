### Introdução ao Pandas

```python
#O método series no pandas cria matrizes unidimensionais, semelhantes a uma coluna do excel ou uma lista do Python.

import pandas as pd
import numpy as np

serie_inicial = pd.Series(np.random.rand(5), index=['A', 'B', 'C', 'D', 'E'])
print(serie_inicial)

serie_lista = pd.Series(['Python', 'Java', 'C++', 'JavaScript'])
print(serie_lista)
```

```python
#A partir de Series
dicionario = {
    "empresa": pd.Series(["Google", "Amazon", "Facebook"], index=['2019', '2020', '2021']),
    "preco": pd.Series([1000, 2000, 3000], index=['2019', '2020', '2021']),
    "volume": pd.Series([100, 200, 300], index=['2019', '2020', '2021'])
}

dataframe = pd.DataFrame(dicionario)
print(dataframe)
```

```python
#DataFrame é uma estrutura de dados bidimensional, semelhante a uma tabela, onde os dados são organizados em linhas e colunas.

#Criando um DataFrame 3x3 a partir de um dicionário

dicionario = {
    "empresa": ["Google", "Amazon", "Facebook"],
    "preco": [1000, 2000, 3000],
    "volume": [100, 200, 300]
}

dataframe = pd.DataFrame(dicionario, index=['2019', '2020', '2021'])
columns = ['empresa', 'preco', 'volume', 'teste']

dataframe
```

```python
#Desafio 2
import random
import matplotlib.pyplot as plt

cotacao_inicial = float(input('Informe a cotação inicial da ação: '))
numero_dias = int(input('Informe o número de dias para simulação: '))
lista_cotacoes = [cotacao_inicial]
lista_dias = [0]
volatilidade = 0.01  # 1% de volatilidade diária

for i in range(1, numero_dias + 1):
    # Gera um retorno diário aleatório baseado na volatilidade
    retorno_diario = np.random.normal(0, volatilidade)
    nova_cotacao = lista_cotacoes[-1] * (1 + retorno_diario)
    lista_cotacoes.append(nova_cotacao)
    lista_dias.append(i)
    print(f'Dia {i}: Cotação = {nova_cotacao:.2f}')


print(lista_cotacoes)

plt.plot(lista_cotacoes)

```
