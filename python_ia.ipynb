```python
def resumo(emails):
    lista_de_resumos = []

    for numero, email in enumerate(emails):
        response = model.generate_content(f"""Vou te mandar o corpo de um email. 
            Qual é o resumo em apenas uma linha, 
            passando o que essa pessoa queria no email, segue o email: {email}""")
        print(f"Email {numero + 1}:", response.text)
        lista_de_resumos.append(f'E-mail {numero + 1}: {response.text}')
        print("-" * 50)
    
    return lista_de_resumos
```

```python
from groq import Groq

# Carrega as variáveis do arquivo .env
load_dotenv()

# Acessar a chave da API do Groq
api_key = os.getenv("GROQ_API_KEY")

client = Groq(api_key=api_key)
completion = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[
      {
        "role": "user",
        "content": "O que é ia?"
      }
    ],
    temperature=1, #0- é o menos criativo possível e 2- é o mais criativo possivel
    max_completion_tokens=8192,
    top_p=1,
    reasoning_effort="medium",
    stream=True,
    stop=None
)

for chunk in completion:
    print(chunk.choices[0].delta.content or "", end="")
```

```python
import csv
import random

# Dados de exemplo
nomes = [
    "Ana Silva", "Bruno Souza", "Carlos Pereira", "Daniela Lima", "Eduardo Castro",
    "Fernanda Alves", "Gabriel Rocha", "Helena Martins", "Igor Mendes", "Julia Santos"
]

# Gera médias aleatórias de 0 a 10
medias = [round(random.uniform(0, 10), 2) for _ in nomes]

# Cria o arquivo CSV
with open('notas.csv', 'w', newline='', encoding='utf-8') as arquivo:
    escritor = csv.writer(arquivo)
    escritor.writerow(["Nome", "Média"])  # Cabeçalho
    for nome, media in zip(nomes, medias):
        escritor.writerow([nome, media])

print("✅ Arquivo 'notas.csv' gerado com sucesso!")
```
