```python
import numpy as np
```

```python
#Lista ou tupla

lista = [1,2,3,4,5]
tupla = (1,2,3,4,5) 

array_lista = np.array(lista)
print(array_lista)    
print(type(array_lista))  

array_tupla = np.array(tupla)
print(array_tupla)    
print(type(array_tupla))
```

```python
#Usando range

array_simples = np.arange(18)
array_sequencial = np.arange(10, 100, 10)

print("Array simples: ", array_simples)
print("Array sequencial: ", array_sequencial)
```
