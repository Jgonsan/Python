```python
import numpy as np
```

```python
#Lista ou tupla

lista = [1,2,3,4,5]
tupla = (1,2,3,4,5) 

array_lista = np.array(lista)
print(array_lista)    
print(type(array_lista))  

array_tupla = np.array(tupla)
print(array_tupla)    
print(type(array_tupla))
```

```python
#Usando range

array_simples = np.arange(18)
array_sequencial = np.arange(10, 100, 10)

print("Array simples: ", array_simples)
print("Array sequencial: ", array_sequencial)
```

```python
#Criando arrays de zeros, uns e vazios

array_zeros = np.zeros((5))
print("Array de zeros: \n", array_zeros)

array_uns = np.ones((5), dtype=np.int16)
print("Array de uns: \n", array_uns)

array_vazio = np.empty((5))
print("Array vazio: \n", array_vazio)
```

```python
#Gerando números aleatórios

vetor_aleatorio = np.random.random((5))
print("Vetor aleatório: \n", vetor_aleatorio)

matriz_aleatoria = np.random.random((3,3))
print("Matriz aleatória: \n", matriz_aleatoria)

```

```python
vetor = np.random.randint(low=5, high=15, size=4)
print("Vetor com números inteiros aleatórios: \n", vetor)
```

```python
vetor = np.random.uniform(low=5, high=15, size=4)
print("Vetor com números uniforme: \n", vetor)
```

```python
#Criando array com mais de uma dimensão
array_2d = np.array([[1,2,3],[4,5,6]])
print("Array 2D: \n", array_2d)
print("Dimensões do array: ", array_2d.shape)
print("Número de dimensões: ", array_2d.ndim)
print("Tipo de dados: ", array_2d.dtype)
print("Tamanho de cada elemento em bytes: ", array_2d.itemsize)
print("Número total de elementos: ", array_2d.size)
print("Tipo do objeto: ", type(array_2d))
```

```python
lista1 = [1,2,3]
lista2 = [4,5,6]
lista3 = [7,8,9]

array_3d = np.array([lista1, lista2, lista3])
print("Array 3D: \n", array_3d)
print("Dimensões do array: ", array_3d.shape)
print("Número de dimensões: ", array_3d.ndim)
print("Tipo de dados: ", array_3d.dtype)
print("Tamanho de cada elemento em bytes: ", array_3d.itemsize)
print("Número total de elementos: ", array_3d.size)
print("Tipo do objeto: ", type(array_3d))

array_2d = np.array([lista1, lista2])
print("Array 2D: \n", array_2d)
print("Dimensões do array: ", array_2d.shape)
print("Número de dimensões: ", array_2d.ndim)
print("Tipo de dados: ", array_2d.dtype)
print("Tamanho de cada elemento em bytes: ", array_2d.itemsize)
print("Número total de elementos: ", array_2d.size)
print("Tipo do objeto: ", type(array_2d))

```

```python
#Criando matrizes

lista1 = [1,2,3]
lista2 = [4,5,6]
lista3 = [7,8,9]

matriz = np.matrix([lista1, lista2, lista3])
print("Matriz: \n", matriz)
print("Dimensões da matriz: ", matriz.shape)
print("Número de dimensões: ", matriz.ndim)
print("Tipo de dados: ", matriz.dtype)
print("Tamanho de cada elemento em bytes: ", matriz.itemsize)
print("Número total de elementos: ", matriz.size)
print("Tipo do objeto: ", type(matriz))

```

```python
#Soma de listas e arrays

lista1 = [1,2,3]
lista2 = [4,5,6]
lista3 = lista1 + lista2
print("Soma de listas: ", lista3)
array1 = np.array(lista1)
array2 = np.array(lista2)
array3 = array1 + array2
print("Soma de arrays: ", array3)
#Subtração de listas e arrays
lista1 = [1,2,3]
lista2 = [4,5,6]
lista3 = [a - b for a, b in zip(lista1, lista2)]
print("Subtração de listas: ", lista3)
array1 = np.array(lista1)
array2 = np.array(lista2)
array3 = array1 - array2
print("Subtração de arrays: ", array3)
```
