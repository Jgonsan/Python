```python
# Notas do(a) estudante:
notas = {'1 Trimestre': 8.5, '2 Trimestre': 9.0, '3 Trimestre': 7.5
         , '4 Trimestre': 8.0}
notas
```

```python
# Calcular a soma

soma = 0 
for nota in notas.values():
    soma += nota
soma
```

```python
Somatorio = sum(notas.values())
Somatorio
```

```python
# Usando a função len() para calcular a média das notas
qtd_notas = len(notas)
media = round(Somatorio / qtd_notas,2)
media

```
